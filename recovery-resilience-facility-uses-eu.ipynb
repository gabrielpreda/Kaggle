# Quick data glimpse

```python
df.geo.value_counts()
```

The field `geo` represents EU countries or EU_27 total.

```python
df.unit.value_counts()
```

The field `unit` has the following meanings:
- MIO_EUR - millions Euro
- PC_ALLOC - percent from allocated value
- PC_GDP - percent from (country) GDP

```python
df.freq.value_counts()
```

The frequence is yearly only.

```python
df.na_item.value_counts()
```

The meaning of `na_item` is:
* CAP — Capital expenditure (all capital outlays).
* CUR — Current expenditure (wages, goods/services, subsidies, interest, etc.).
* TOTAL — Total expenditure = CUR + CAP.
* P51G — Gross fixed capital formation (investment in fixed assets like buildings, machinery).
* D9 — Capital transfers (e.g., investment grants, other one-off capital transfers).
* OTH — Other (a residual/“other” bucket; often for items not elsewhere classified).
* CAP_OTH — Other capital expenditure (capital items other than P.51g and D.9).

```python
df.fin_typ.value_counts()
```

Tne `fin_typ` (financing type) can be:
- TOTAL - total i.e. grant & loan
- GRNT - grant
- LOAN - loan

```python
df.TIME_PERIOD.value_counts()
```

The `TIME_PERIOD` represents the year.
