```python
import kaggle_benchmarks as kbench

@kbench.task(name="family_relationship_riddle")
def family_relationship_riddle(llm):
    prompt = "My sister Anna will marry my cousin second removed. What is her father to me?"
    response = llm.prompt(prompt)

    kbench.assertions.assert_contains_regex(
        r"(?i)father",
        response,
        expectation="The model should correctly identify that my sister's father is my father."
    )

family_relationship_riddle.run(kbench.llm)
```
