# Embeddings with PyTorch

```python
import torch
import torch.nn as nn

num_categories = 1000
embedding_dim = 8

embedding = nn.Embedding(num_categories, embedding_dim)

# Example batch of categorical indices
category_indices = torch.tensor([49, 31, 521, 312])

embedded_vectors = embedding(category_indices)

print(embedded_vectors.shape)  # (batch_size, embedding_dim)

print(embedded_vectors) # the actual embeddings
```

# FeatureHasher 

```python
from sklearn.feature_extraction import FeatureHasher

hasher = FeatureHasher(n_features=8, input_type='string')

data = [['Lancia'], ['Audi'], ['Porche'], ['Lancia']]
hashed = hasher.transform(data)

print(hashed.toarray())
```
