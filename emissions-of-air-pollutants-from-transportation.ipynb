# Introduction

Emissions from the transport sector is the sum of emissions from road traffic, shipping, aviation, rail traffic and mobile machinery.   
Road traffic has the largest share of emissions from transport, especially from passenger cars.   

## Dataset

The dataset is: [Belgium emissions of air pollutants from transport](https://www.kaggle.com/datasets/gpreda/belgium-emissions-of-air-pollutants-from-transport).

## Columns meaning

The meaning of the column names (in Dutch) is (translated in English):

* **mobiele machines** → mobile machinery
* **spoor niet-uitlaat** → rail non-exhaust (e.g. wear from brakes, rails)
* **spoor uitlaat** → rail exhaust (emissions from locomotives, engines)
* **luchtvaart** → aviation
* **scheepvaart** → shipping / maritime transport
* **wegverkeer niet-uitlaat** → road traffic non-exhaust (e.g. brake, tire, road wear)
* **wegverkeer uitlaat** → road traffic exhaust (tailpipe emissions)

# Data exploration

```python
plt.figure(figsize=(15,6))

for col in df.columns[1:]:
    plt.plot(df["Data"], df[col], marker="o", label=col)

plt.title("Emissions by source category (2005–2009)")
plt.xlabel("Year")
plt.ylabel("Value")
plt.grid(True, linestyle="--", alpha=0.6)
plt.legend(title="Category", bbox_to_anchor=(0.83, 1), loc="upper left")
plt.tight_layout()
plt.show()
```

Let's show the y-scale with log.

Let's translate the column names in English (from Dutch) for easier readability.

```python
df.columns = ["Year", "Mobile machinery", "Rail non-exhaust", "Rail exhaust", "Aviation", 
              "Shipping/maritime transport", "Road traffic non-exhaust", "Road traffic exhaust"]
```

```python
df["Year"] = pd.to_datetime(df["Year"], format="%Y")
```
