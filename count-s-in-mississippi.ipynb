```python
import kaggle_benchmarks as kbench
import re

@kbench.task(name="count 's' in mississippi")
def count_s_in_mississippi(llm):
    response = llm.prompt("How many letters \"s\" are in \"mississippi\"?")
    kbench.assertions.assert_contains_regex(
        r"(?i)\b(4|four)\b",
        response,
        expectation="The model should answer that there are 4 's's."
    )

count_s_in_mississippi.run(kbench.llm)
```
