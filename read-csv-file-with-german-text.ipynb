# Introduction

The data is about criminal activity in 2023 in Germany. It contains criminal activities stats, grouped by victim type.

```python
import pandas as pd
```

```python
path = "/kaggle/input/2023-polizeiliche-kriminalstatistik-deutschland/ST-O-01-T91-Staedte-Opfer_csv.csv"
```

```python
def load_german_csv(p, try_encodings=("utf-8-sig","latin1","cp1252")):
    last_err = None
    for enc in try_encodings:
        for params in (
            dict(sep=None, engine="python"),       # sniff
            dict(sep=";"),                         # force semicolon
        ):
            try:
                print(f"Encoding: {enc} Params: {params}")
                return pd.read_csv(p, encoding=enc, **params)
            except Exception as e:
                last_err = e
    raise last_err

df = load_german_csv(path)
```
