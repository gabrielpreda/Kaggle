```python
import kaggle_benchmarks as kbench
import re

@kbench.task(name="machine_riddle_task")
def machine_riddle_task(llm):
    prompt = "If it takes 5 machines 5 minutes to make 5 items, how long would it take 100 machines to make 100 items?"
    response = llm.prompt(prompt)

    kbench.assertions.assert_contains_regex(
        r"\b5\b|\bfive\b",
        response,
        expectation="The answer should be 5 minutes.",
    )

machine_riddle_task.run(kbench.llm)
```
