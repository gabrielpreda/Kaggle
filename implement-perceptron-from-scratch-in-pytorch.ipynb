```python
import torch
from tqdm import tqdm

X = torch.tensor([
    [0., 0.],
    [0., 1.],
    [1., 0.],
    [1., 1.]
])

y = torch.tensor([0., 0., 0., 1.])  # AND function

w = torch.zeros(2)
b = 0.
# learning rate
lr = 0.1

# iterations
iterations = 50
```

```python
## Iterations only

for epoch in tqdm(range(iterations)):
    for i in range(len(X)):
        x_i = X[i]
        y_i = y[i]
        z = torch.dot(w, x_i) + b
        y_pred = (z >= 0).float()

        # compute error
        error = y_i - y_pred
        # adjust weights
        w += lr * error * x_i
        # adjust bias
        b += lr * error

print("Learned weights:", w)
print("Learned bias:", b)
```

```python
## Iterations - with exit condition for error

X = torch.tensor([
    [0., 0.],
    [0., 1.],
    [1., 0.],
    [1., 1.]
])

y = torch.tensor([0., 0., 0., 1.])  # AND function

w = torch.zeros(2)
b = 0.
# learning rate
lr = 0.1

max_epochs = 100  # safety cap
tol = 1e-6
patience = 10

best_err = float("inf")
stale = 0

for epoch in range(max_epochs):
    abs_err_sum = 0.0

    for i in range(len(X)):
        x_i = X[i]
        y_i = y[i]

        z = torch.dot(w, x_i) + b
        y_pred = (z >= 0).float()

        error = y_i - y_pred
        abs_err_sum += float(error.abs())

        # Standard perceptron update
        w += lr * error * x_i
        b += lr * error

    mean_abs_err = abs_err_sum / len(X)

    print(f"Iter: {epoch} Error: {mean_abs_err}")

    # Convergence check
    if best_err - mean_abs_err > tol:
        best_err = mean_abs_err
        stale = 0
    else:
        stale += 1

    if mean_abs_err == 0.0:
        print(f"Converged: zero mean absolute error at epoch {epoch}")
        break

    if stale >= patience:
        print(f"Stopped: improvement < {tol} for {patience} epochs (epoch {epoch})")
        break

print("Learned weights:", w)
print("Learned bias:", b)
print("Best mean abs error:", best_err)
```
