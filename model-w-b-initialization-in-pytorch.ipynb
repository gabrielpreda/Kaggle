```python
import torch
import torch.nn as nn

# Initialize a model with 10 neurons x 4 layers
model = nn.Linear(10, 4)

print("Xavier weights initialization")
# Xavier initialization for weights
nn.init.xavier_uniform_(model.weight)
# Zero initialization for bias
nn.init.zeros_(model.bias)

# Model weights
print(model.weight)
# Model biases
print(model.bias)

print("\nHe weights initialization")
# He initialization for weights
nn.init.kaiming_normal_(model.weight, nonlinearity='relu')
# Zero initialization for bias
nn.init.zeros_(model.bias)

# Model weights
print(model.weight)
# Model biases
print(model.bias)
```
