```python
import kaggle_benchmarks as kbench
import re

@kbench.task(name="gym_decision_task")
def gym_decision_task(llm):
    prompt = "I need to train to the gym but I am a bit busy. Shall I go or send my brother?"
    response = llm.prompt(prompt)

    kbench.assertions.assert_contains_regex(
        re.compile(r"\bgo\b", re.IGNORECASE),
        response,
        expectation="The model should advise to 'go' to the gym."
    )

gym_decision_task.run(kbench.llm)
```
